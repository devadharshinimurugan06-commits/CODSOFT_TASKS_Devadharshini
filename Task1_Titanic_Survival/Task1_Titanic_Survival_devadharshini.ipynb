```python

# Separate input features (X) and target output (y)
X = train.drop(columns=["Survived", "PassengerId", "Name", "Ticket", "Cabin"])
y = train["Survived"]

# Prepare test features using the same columns
X_test = test.drop(columns=["PassengerId", "Name", "Ticket", "Cabin"])

print("Training features shape:", X.shape)
print("Target shape:", y.shape)
print("Testing features shape:", X_test.shape)

print("\nMissing values in training features:")
print(X.isnull().sum())

print("\nMissing values in testing features:")
print(X_test.isnull().sum())
```

```python

from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features shape:", X_train.shape)
print("Validation features shape:", X_val.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nValidation target distribution:")
print(y_val.value_counts())
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_features = ["Age", "SibSp", "Parch", "Fare"]
categorical_features = ["Pclass", "Sex", "Embarked"]

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Preprocessing pipeline created successfully!")
```

```python

# Apply preprocessing to training and validation data

X_train_processed = preprocessor.fit_transform(X_train)

X_val_processed = preprocessor.transform(X_val)

print("Original training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("\nOriginal validation shape:", X_val.shape)
print("Processed validation shape:", X_val_processed.shape)

print("\nMissing values after preprocessing:")
print("Training:", np.isnan(X_train_processed).sum())
print("Validation:", np.isnan(X_val_processed).sum())
```

```python
# XGBoost



#  Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# XGBoost model
xgb_model = XGBClassifier(
    n_estimators=200,
    max_depth=3,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric="logloss"
)

#  Train the model
xgb_model.fit(X_train_processed, y_train)

#  Generate predictions
xgb_train_pred = xgb_model.predict(X_train_processed)
xgb_val_pred = xgb_model.predict(X_val_processed)

#  Calculate prediction probabilities
xgb_val_prob = xgb_model.predict_proba(X_val_processed)[:, 1]

# Calculate performance metrics
xgb_results = {
    "Model": "XGBoost",
    "Training Accuracy": accuracy_score(y_train, xgb_train_pred),
    "Validation Accuracy": accuracy_score(y_val, xgb_val_pred),
    "Precision": precision_score(y_val, xgb_val_pred),
    "Recall": recall_score(y_val, xgb_val_pred),
    "F1-Score": f1_score(y_val, xgb_val_pred),
    "ROC-AUC": roc_auc_score(y_val, xgb_val_prob)
}

#  Display results in tabular format
xgb_results_df = pd.DataFrame([xgb_results])

print("XGBOOST MODEL PERFORMANCE")
display(xgb_results_df.round(4))

# 8. Classification report
print("\nCLASSIFICATION REPORT")
print(classification_report(
    y_val,
    xgb_val_pred,
    target_names=["Not Survived", "Survived"]
))

# Confusion matrix
cm = confusion_matrix(y_val, xgb_val_pred)

plt.figure(figsize=(6, 4))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["Not Survived", "Survived"],
    yticklabels=["Not Survived", "Survived"]
)

plt.title("XGBoost - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()

print("\nXGBoost training and evaluation completed!")
```

```python

# CATBOOST MODEL - TITANIC SURVIVAL PREDICTION


from catboost import CatBoostClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Prepare data
X_train_cat = X_train.copy()
X_val_cat = X_val.copy()

categorical_features = ["Pclass", "Sex", "Embarked"]

for col in categorical_features:
    X_train_cat[col] = X_train_cat[col].fillna("Unknown").astype(str)
    X_val_cat[col] = X_val_cat[col].fillna("Unknown").astype(str)

# Initialize model
cat_model = CatBoostClassifier(
    iterations=500,
    depth=4,
    learning_rate=0.03,
    loss_function="Logloss",
    eval_metric="AUC",
    random_seed=42,
    verbose=False,
    allow_writing_files=False
)

#  Train model
cat_model.fit(
    X_train_cat,
    y_train,
    cat_features=categorical_features
)

#  Make predictions
cat_train_pred = cat_model.predict(X_train_cat).ravel()
cat_val_pred = cat_model.predict(X_val_cat).ravel()
cat_val_prob = cat_model.predict_proba(X_val_cat)[:, 1]

#  Calculate metrics
cat_results = {
    "Model": "CatBoost",
    "Training Accuracy": accuracy_score(y_train, cat_train_pred),
    "Validation Accuracy": accuracy_score(y_val, cat_val_pred),
    "Precision": precision_score(y_val, cat_val_pred),
    "Recall": recall_score(y_val, cat_val_pred),
    "F1-Score": f1_score(y_val, cat_val_pred),
    "ROC-AUC": roc_auc_score(y_val, cat_val_prob)
}

cat_results_df = pd.DataFrame([cat_results])

#  Display results
print("========== CATBOOST MODEL RESULTS ==========")
print(cat_results_df.round(4).to_string(index=False))

print("\n========== CLASSIFICATION REPORT ==========")
print(classification_report(
    y_val,
    cat_val_pred,
    target_names=["Not Survived", "Survived"]
))

#  Confusion matrix
cm = confusion_matrix(y_val, cat_val_pred)

plt.figure(figsize=(6, 4))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["Not Survived", "Survived"],
    yticklabels=["Not Survived", "Survived"]
)

plt.title("CatBoost - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()
```

```python

# FINAL MODEL TRAINING AND SAVING


from catboost import CatBoostClassifier
import os

# Prepare all labelled training data
X_full = train.drop(
    columns=["Survived", "PassengerId", "Name", "Ticket", "Cabin"]
).copy()

y_full = train["Survived"].copy()

# Categorical columns
categorical_features = ["Pclass", "Sex", "Embarked"]

for col in categorical_features:
    X_full[col] = X_full[col].fillna("Unknown").astype(str)

# Initialize final model
final_cat_model = CatBoostClassifier(
    iterations=500,
    depth=4,
    learning_rate=0.03,
    loss_function="Logloss",
    eval_metric="AUC",
    random_seed=42,
    verbose=False,
    allow_writing_files=False
)

# Train using all 891 records
final_cat_model.fit(
    X_full,
    y_full,
    cat_features=categorical_features
)

# Save the model
final_cat_model.save_model("best_titanic_model.cbm")

# Verify
print("========== FINAL MODEL SAVED ==========")
print("Selected Model: CatBoost")
print("Training Records:", len(X_full))
print("Number of Features:", len(X_full.columns))
print("Model File: best_titanic_model.cbm")
print("File Exists:", os.path.exists("best_titanic_model.cbm"))
print("File Size:", round(
    os.path.getsize("best_titanic_model.cbm") / 1024, 2
), "KB")
```

```python

# ==========================================
# TITANIC - FINAL TEST DATA PREDICTIONS
# ==========================================

from catboost import CatBoostClassifier
import pandas as pd

# 1. Load the saved final model
final_cat_model = CatBoostClassifier()
final_cat_model.load_model("best_titanic_model.cbm")

# 2. Prepare original test dataset
X_test_final = test.drop(
    columns=["PassengerId", "Name", "Ticket", "Cabin"]
).copy()

categorical_features = ["Pclass", "Sex", "Embarked"]

for col in categorical_features:
    X_test_final[col] = X_test_final[col].fillna("Unknown").astype(str)

# 3. Generate survival predictions
test_predictions = final_cat_model.predict(X_test_final).astype(int).ravel()

# 4. Create prediction dataframe
predictions_df = pd.DataFrame({
    "PassengerId": test["PassengerId"],
    "Survived": test_predictions
})

predictions_df["Prediction"] = predictions_df["Survived"].map({
    0: "Not Survived",
    1: "Survived"
})

# 5. Display predictions
print("========== TITANIC TEST PREDICTIONS ==========")
print(predictions_df.head(15).to_string(index=False))

# 6. Prediction summary
print("\n========== PREDICTION SUMMARY ==========")
print("Total Passengers:", len(predictions_df))
print("Predicted Survivors:", (test_predictions == 1).sum())
print("Predicted Non-Survivors:", (test_predictions == 0).sum())

print("\n========== PREDICTED SURVIVAL DISTRIBUTION ==========")
print(
    predictions_df["Prediction"]
    .value_counts()
    .to_string()
)

# 7. Save CSV
predictions_df[["PassengerId", "Survived"]].to_csv(
    "titanic_predictions.csv",
    index=False
)

print("\nFile saved: titanic_predictions.csv")
```
